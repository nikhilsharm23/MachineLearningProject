### Text Preproceesing

```python
# What steps we are performs in this Projects

#covert to lowercase 
#remove URLS
# Remove HTML tags
# remove Stopwords
# Remove Punctuations (? > , # $ )
# Perform Steamming  
# Target value convert into num 
# Perform Vectorizations (TF - IDF) - covert text into Numbers 
```

```python
df['review'] = df['review'].str.lower()
```

```python
# removing urls using regEx 
import re

def remove_urls(text):
    return re.sub(r"http\s+","",text)

df['review'] = df['review'].apply(remove_urls)
```

```python
# removing punctuations 
def remove_punctuations(text):
    text = re.sub(r"[^A-Za-z0-9\s]","",text)
    return text
```

```python
df['review'] = df['review'].apply(remove_punctuations)
```

```python
# remove HTML Tags
```

```python
def remove_html(text):
    text = re.sub(r"<.*?>","",text)
    return text
df['review'] = df['review'].apply(remove_html)
```

```python
# Removing Stop Words 
```

```python
import nltk 
nltk.download("punkt")

nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
```

```python
# remove stop words 

def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words("english")

    for word in tokens:
        if word in stop_words:
            text.replace(word , "")
    return text
df['review'] = df['review'].apply(remove_stopwords)
```

```python
df.head()
```

```python
from nltk.stem import PorterStemmer 

def stemming(text):
    ps = PorterStemmer()
    stemmed_words = []

    tokens = word_tokenize(text)
    for token in tokens:
        stemmed_token = ps.stem(token)
        stemmed_words.append(stemmed_token)
    return " ".join(stemmed_words)
df['review'] = df['review'].apply(stemming)
```

```python
import torch 
from torch.utils.data import TensorDataset , DataLoader 
```

```python
X_train = X_train.toarray()
X_test = X_test.toarray()
```

```python
train_set = TensorDataset(
    torch.from_numpy(X_train).float(),
    torch.from_numpy(y_train.values).float()
)

test_set = TensorDataset(
    torch.from_numpy(X_test).float(),
    torch.from_numpy(y_test.values).float()
)
```

```python
train_loader = DataLoader(train_set , shuffle=True , batch_size=64)
test_loader = DataLoader(test_set , shuffle=True , batch_size=64)
```

# Build RNN

```python
import torch
import torch.nn as nn
import torch.optim as optim

class RNN(nn.Module):
    def __init__(self, input_size, hidden_size=128, num_layers=1):
        super(RNN, self).__init__()

        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # RNN layer
        self.rnn = nn.RNN(
            input_size,
            hidden_size,
            num_layers,
            batch_first=True
        )

        # Fully connected layer
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        batch_size = x.size(0)

        h0 = torch.zeros(
            self.num_layers,
            batch_size,
            self.hidden_size
        ).to(x.device)   # important fix

        out, _ = self.rnn(x, h0)

        out = self.fc(out[:, -1, :])  # last timestep
        return out
```

```python
input_size = X_train.shape[1]
model  = RNN(input_size)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters())
```

# Training the RNN

```python
epochs = 10
for epoch in range(epochs):
    model.train()

    for Xb , yb in train_loader:
        optimizer.zero_grad()

        Xb = Xb.unsqueeze(1)

        outputs = model(Xb)  
        outputs = torch.sigmoid(outputs.squeeze())

        loss = criterion(outputs , yb)

        loss.backward()  
        optimizer.step()  # weights update

    print(f'epoch = {epoch+1}/{epoch} and loss = {loss.item()}')
```

```python
model.eval()

with torch.no_grad():
    correct_vals = 0
    tot_vals = 0

    for Xb , yb in test_loader:
        Xb = Xb.unsqueeze(1)
        outputs = model(Xb)
        predicted = (torch.sigmoid(outputs.squeeze()) > 0.5).float()
        tot_vals += yb.size(0)
        correct_vals += (predicted == yb).sum().item()

    print(f'accuracy = {correct_vals/tot_vals*100}')
```
